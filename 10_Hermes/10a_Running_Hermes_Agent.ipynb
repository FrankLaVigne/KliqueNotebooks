# 10a Running Hermes Agent

**Purpose:** Install and configure Hermes Agent on top of Klique, using a model served through the LLM Gateway.

**What we're doing**

Confirm an LLM Gateway endpoint, launch Hermes Agent from Klique Applications, explore its terminal UI, and stop the session.

**Why we're doing it**

[Hermes Agent](https://hermes-agent.nousresearch.com/) is an open-source, self-hosted AI agent framework built by Nous Research. It runs continuously to automate tasks, remembers context across sessions, and writes its own reusable skills over time. On Klique it runs as an application on managed compute, pointed at a model endpoint you already govern.

**What to look for**

* An LLM Gateway endpoint pointing at a vLLM-hosted model
* Hermes Agent running in a VS Code terminal inside the app
* The model Hermes reports with `/model`
* A workspace that persists between sessions

**Output:** A Hermes Agent session that uses your LLM Gateway endpoint, explored and then stopped.

This section concerns **Hermes Agent**, not a Hermes model or the ClearML worker agent.

## 10a.1 Prerequisites: an LLM Gateway Endpoint

Hermes needs a model to talk to. Before launching it, make sure an endpoint is defined in **LLM Gateway**.

This walkthrough uses a single endpoint named **Hermes endpoint**. It points to a vLLM instance running on the same cluster (see [Deploy a Model with vLLM](../01_vLLM/01_Deploy_a_Model_with_vLLM.ipynb)), serving `Qwen/Qwen3.5-2B`. That small model is good enough for this use case.

Open the endpoint and check its details: the endpoint URL and the model name are configured in the LLM Gateway. As the endpoint view and diagram show, there is no smart routing here. It's a basic setup, just to get started.

You also need a Klique login with access to the Hermes application, a compute queue, and a project.

> **Checkpoint:** LLM Gateway lists an endpoint (here, **Hermes endpoint**) that points at a running vLLM model.

## 10a.2 Configure and Launch

Go to **Applications** and select **Hermes**. Fill in the launch form:

| Field | Value used in this walkthrough |
|---|---|
| **Name** | `Hermes Agent` |
| **Compute Resource** | Your default queue |
| **Project** | Your project |
| **Model** | `Hermes endpoint`, the LLM Gateway endpoint from 10a.1 |
| **Update Hermes at session start** | Checked |

Leave everything else at the defaults, click **Launch**, and wait for the app to spin up.

> **Screenshot placeholder:** Hermes launch form with name, default queue, project, Hermes endpoint selected as the model, and Update Hermes at session start checked.

<!-- screenshot: images/hermes_launch.png (Hermes launch form before submission; do not include credentials.) -->

## 10a.3 Open the App

Once the app is running, click **Open App**. The UI opens an instance of VS Code, straight into a terminal with Hermes running.

Because **Update Hermes at session start** is checked, Hermes first checks for updates. Wait for that to complete.

The first question is **Help improve Hermes**: share data with Nous Research, or **No thanks**. This walkthrough chooses **No thanks** and presses Enter. Follow your organization's policy here.

Hermes then restarts to finish the update, and the terminal UI comes up.

> **Checkpoint:** The Hermes terminal UI is up and shows the model you configured (here, Qwen3.5-2B).

> **Screenshot placeholder:** VS Code terminal inside the app showing the Hermes terminal UI after the update, with the configured model visible.

## 10a.4 Ask the Agent a Question

Ask the agent a question, for example:

```
What changed in this repo lately?
```

The agent answers that the workspace is not set up as a Git repo in the current implementation. It still finds what is there: an `agents.md` file, a `.vscode` folder, and other information about the workspace.

> **Checkpoint:** The agent responds using your endpoint and describes the workspace.

> **Screenshot placeholder:** The question and the agent's answer about the workspace.

## 10a.5 Explore the Terminal UI

Try a few commands:

* Type `help` to see all the options available.
* Choose `gateway` if you want to set up messaging, along with the other options available to a Hermes Agent setup.
* Type `/model` to see which model you're running. Here it shows Qwen3.5-2B, local, on a custom endpoint.

From `/model` you can also set the reasoning effort. It starts at **medium**; select **high**, and the reasoning effort is now set to high.

> **Checkpoint:** `/model` shows your endpoint's model, and you can change the reasoning effort.

## 10a.6 Examine `agents.md`

Open the `agents.md` file in the workspace. It holds the instructions for the agent and how it is set up.

A full introduction to Hermes Agent and everything it can do is beyond the scope of this lesson. The point here is that you can run it inside Klique.

The workspace directory is preserved between sessions. Any changes you make to `agents.md`, and any directories or files you add, are still there across multiple sessions.

## 10a.7 Stop the Session

If you're continuing to the next lesson in the series, you can leave the session running.

To save cost and resources, click **Stop Session** to stop the application instance. You are asked to confirm; click **Stop**, and the system shuts the application down automatically.

> **Checkpoint:** The Hermes Agent instance is stopped.

## Troubleshooting

| Symptom | What to check |
|---|---|
| Hermes is missing from Applications | Ask your administrator about catalog access. |
| `Hermes endpoint` is not in the Model list | Confirm the endpoint exists in LLM Gateway and that you can use it. |
| The app stays on setup | Check queue availability and the app's console output. |
| Hermes sits on the update after opening | Wait for the update and restart to finish before typing. |
| Authentication or model error | Confirm the vLLM deployment is running and the endpoint's URL and model name are correct. |
| Files missing after a restart | Check that they were saved in the workspace directory, which is what persists. |

## Wrap-Up

You pointed an LLM Gateway endpoint at a vLLM-hosted model, launched Hermes Agent from Klique Applications, and explored its terminal UI. Your workspace persists between sessions, so stop the session anytime to save resources.
