# 0 Orientation

*Set the scope and the mental model before you deploy anything.*

## 0.1 What This Lab Is (and Is Not)

This lab **is**:

- A guided walkthrough of running vLLM **on Klique**: deploying a model, getting an endpoint, and putting that endpoint under governance
- Focused on the Klique objects you work with: the vLLM Model Deployment app, Model Endpoints, the AI Application Gateway, and the LLM Gateway
- One model, one GPU, one path from start to finish

This lab is **not**:

- A vLLM tutorial. Klique runs vLLM for you; you will not tune engine arguments or write a serving config.
- A Kubernetes or GPU setup guide. Your administrator has already set up the queues and agents.
- A Klique installation or administration guide

> **In one sentence:** "Klique turns *a model name* into *a governed endpoint*. This lab shows every step in between."


## 0.2 The Mental Model

Every model you serve on Klique follows the same path:

```
Model  →  vLLM Model Deployment  →  Model Endpoint  →  AI Application Gateway  →  LLM Gateway
 (what)     (the app instance          (the URL you      (authenticates every      (decides who can use
             that runs vLLM)            call)             request)                  it, and tracks cost)
```

| Step | What it is | Where you see it | Section |
|---|---|---|---|
| **Model** | A HuggingFace model name, or a model registered in Klique | The launch form's *Model* field | 1 |
| **vLLM Model Deployment** | A Klique app. Each launch creates an *instance* that runs vLLM on a GPU from a queue. | Applications → vLLM Model Deployment | 1 |
| **Model Endpoint** | The URL the instance serves | [Model Endpoints](https://klique.ai/docs/latest/docs/webapp/webapp_model_endpoints) | 1 |
| **AI Application Gateway** | The secure, authenticated route to the endpoint. Each request carries a gateway token. | Settings → AI Application Gateway Tokens | 1 |
| **LLM Gateway** | Klique's central point of control for LLM traffic: who can reach which model, budgets, rate limits, and usage | [LLM Gateway](https://klique.ai/docs/latest/docs/whats_klique/llm_gateway) | 2 |

Key ideas to land:

- You **deploy** a model with an app, not by writing a server
- You **call** it through a gateway, never directly
- You **govern** it centrally, so access and cost are not left to each client


## 0.3 What Success Looks Like

By the end of the lab you should be able to:

- Deploy a model with the vLLM Model Deployment app and explain what each required field does
- Tell when an endpoint is ready, using the instance dashboard and Model Endpoints
- Call the endpoint with a gateway token, and explain why that token is not your Klique login
- Explain what the LLM Gateway adds on top of a working endpoint

Success is **not**:

- Tuning vLLM for throughput
- Memorizing every field on the launch form


## 0.4 How the Lab Runs

- The lab is a set of **guided Jupyter notebooks**, one folder per section, run in order
- Steps in the **Klique WebApp** and **code cells** are interleaved: you click to deploy, then run code to verify
- You are **not expected to write code from scratch**
- Each section ends with a **Next** pointer to the section that builds on it

| Section | Directory | Topic |
|---|---|---|
| 00 | `00_Orientation/` | This notebook: scope, mental model, setup |
| 01 | `01_vLLM/` | Deploy a model with vLLM on Klique and test its endpoint |
| 02 | `02_LLMGateway/` | Call the endpoint through the LLM Gateway under a profile, then call different models and Routed Resources |


## 0.5 Ground Rules

**One path forward.** One small model on one GPU, with the launch form's defaults wherever possible.

**Questions are welcome; tangents are parked.** vLLM tuning, multi-GPU, and LoRA are real topics, but not today's.

**Credentials stay out of notebooks.** Every secret lives in one `.env` file at the project root, which git ignores. No cell ever prints a full token.


## 0.6 Setting Up

### 0.6.1 What You Need

| Item | Where it comes from |
|---|---|
| A Klique WebApp login | Your administrator |
| A queue with a GPU agent | Your administrator. You pick it as the *Compute Resource* in Section 1. |
| A project you can launch into | Your administrator, or one you create |
| A Jupyter environment that can reach the endpoint's network | Your laptop or a Klique workspace. Model endpoint host names are often internal. |
| Python 3 with JupyterLab | Each notebook installs its own packages |

### 0.6.2 Pull Down the Repo

Open a terminal and clone the lab repository:

```
git clone https://github.com/FrankLaVigne/KliqueNotebooks.git
cd KliqueNotebooks
```

### 0.6.3 About the `.env` File

All sections read their settings from **one** `.env` file at the project root, through `config.py`:

```
KliqueNotebooks/
├── config.py        # loads .env and exposes the settings to every notebook
├── .env             # your endpoint and token (not tracked in git)
├── 00_Orientation/
├── 01_vLLM/
└── 02_LLMGateway/
```

You create `.env` in Section 1.6, once you have an endpoint and a token. Every notebook after that picks it up automatically.

Once the repo is cloned, the environment is ready. Move to Section 1.
