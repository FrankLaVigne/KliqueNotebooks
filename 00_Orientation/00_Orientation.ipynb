# 0 Orientation

*Choose a path and understand the moving parts before you deploy a model or launch a developer session.*

## 0.1 What This Lab Covers

This repository is a hands-on guide to two connected ways of working on Klique:

- **Endpoint labs, Sections 01–04:** deploy a model with vLLM, call it through the AI Application Gateway, add LLM Gateway routing and governance, and run a small tool-using agent.
- **Developer sessions, Sections 05–09:** work on managed compute through JupyterLab, SSH, VS Code, Claude Code, and Claude Desktop. Follow-on labs build applications, test and debug agents, compare models, resume training, and install ComfyUI.

The endpoint path starts with one small model on one GPU, then expands to multiple models and Routed Resources. The developer-session introductions can be taken independently; you do not need to deploy a vLLM model first. Some follow-on exercises use the governed gateway from Section 02.

This is not a vLLM tuning tutorial or a Klique installation guide. Your administrator provides the permitted apps, queues, agents, and environments. You use those resources, inspect what happened, save your work, and stop what you launched.

> **In one sentence:** Klique turns a model name into a governed endpoint and gives you managed workspaces to build around it.

## 0.2 The Mental Model

### Endpoint path: Sections 01–04

The vLLM deployment in this lab follows this path:

```text
Model → vLLM Model Deployment → Model Endpoint → AI Application Gateway → LLM Gateway
 (what)    (the app instance       (the URL you      (authenticates         (routes, meters,
            that runs vLLM)         call)            each request)         and governs)
```

| Step | What it is | Where you see it | Section |
|---|---|---|---|
| **Model** | A model from [vLLM Recipes](https://recipes.vllm.ai/), with its tested vLLM settings | The launch form's *vLLM Arguments* field | 01 |
| **vLLM Model Deployment** | A Klique app. Each launch creates an *instance* that runs vLLM on a GPU from a queue. | Applications → vLLM Model Deployment | 01 |
| **Model Endpoint** | The URL the instance serves | [Model Endpoints](https://klique.ai/docs/latest/docs/webapp/webapp_model_endpoints) | 01 |
| **AI Application Gateway** | The secure, authenticated route to the endpoint. Each request carries a gateway token. | Settings → AI Application Gateway Tokens | 01 |
| **LLM Gateway** | Central control for model access, routing, budgets, rate limits, and usage | [LLM Gateway](https://klique.ai/docs/latest/docs/whats_klique/llm_gateway) | 02–04 |

You **deploy** with an app, **call** through an authenticated gateway, and **govern** access and cost centrally. Section 02 also covers profile aliases and Resources for other models, including external providers; those do not all require your own vLLM deployment. Section 03 makes usage and limits visible, and Section 04 applies them to an agent's repeated model calls.

### Developer-session path: Sections 05–09

```text
Browser or local SSH client → managed developer session → repository, tools, and saved workspace
                                              └──────→ optional model calls through the LLM Gateway
```

The browser or SSH client is your interface; the development environment runs on the selected managed compute. A developer session and a model deployment have **separate lifecycles**. Stopping one does not stop the other. Workspace persistence restores files, not a running Python process or an automatically resumed training job.

The interface also matters: Section 08 uses Claude Code in a remote VS Code environment; Section 09 opens Claude Desktop through a browser-accessible remote desktop. Follow each app's documented account/provider setup rather than assuming they use the same configuration or governance route.

## 0.3 What Success Looks Like

Choose the outcomes that match your path:

- **Endpoints:** deploy a model, recognize when its endpoint is ready, and call it with a gateway token that is distinct from your Klique login.
- **Routing and governance:** call allowed profile aliases, inspect usage, and observe rate-limit and budget enforcement with an administrator's help.
- **Agents:** trace a bounded tool-using agent and see how it stops or handles gateway rejections.
- **Developer sessions:** launch the right interface, confirm where commands run, preserve the workspace, and stop the session.
- **Hands-on projects:** build a governed playground, retrieve repository documentation, debug an agent, compare model responses, resume a training checkpoint, or generate an image with ComfyUI.

You are not expected to memorize every launch-form field or tune vLLM for throughput. Use the checkpoints to distinguish what you observed from what remains deployment-specific.

## 0.5 Ground Rules

**Start small.** Use the lab's bounded prompts, local fixtures, and practice files before live calls. Keep vLLM tuning, multi-GPU serving, and LoRA outside this introductory path.

**Use permitted resources.** Confirm the app, queue, project, image, and network route before launching. Section 03's settings changes and parts of Sections 02b and 04 need administrator access or help. Budget exercises can lock a user out for the rest of the month; use the throwaway user or project described in those notebooks and know how to restore the settings.

**Keep credentials private.** The endpoint labs use an ignored root `.env`; the VS Code gateway helper also supports approved injected environment variables. Follow the Claude apps' separate account/provider setup. Do not put credentials in notebook output, source, screenshots, or Git. An ignored file can still be included in a workspace snapshot, so review snapshot access and secret provisioning.

**Save, then stop.** Source control, workspace snapshots, and training checkpoints solve different problems. Keep the work you need, record session IDs when restoring later, and stop your developer sessions, servers, and separately owned model/GPU instances as each notebook directs.

For either path, have a Klique WebApp login, permission to use the selected app, an approved project and compute queue, and administrator guidance on the environment and network access.

**Package setup is not automatic.** The current endpoint notebooks import `requests` but do not install it. Prepare your Jupyter environment through an approved package source before starting Section 01; for example, in that environment's terminal:

### 0.6.3 Settings and Repository Layout

**Endpoint notebooks 01–04** read the root `.env` through `config.py`. Create it in Section 1.6 after obtaining the endpoint and gateway token:

```text
KLIQUE_BASE_URL=https://<your endpoint>/v1
KLIQUE_MODEL=<your Model Endpoint Name>
KLIQUE_GATEWAY_TOKEN=<your AI Application Gateway token>
```

For this endpoint path, `config.py` derives `KLIQUE_LLM_GATEWAY_URL` as `https://<same-host>/llm`. Set the profile and gateway alias names in the notebooks. The optional API-based model listing in 02b uses additional Klique API credentials; the manual alias list avoids that optional setup.

**VS Code live gateway projects 07b–07e** use their own `gateway_client.py`, not root `config.py`. They can reuse the root `.env` or approved injected variables, and additionally need `KLIQUE_PROFILE` and `KLIQUE_MODEL_ALIASES`. Those aliases are names in the profile, not necessarily the direct endpoint's `KLIQUE_MODEL`. See [gateway.env.example](../07_VSCode/gateway.env.example). The helper also supports an explicit `KLIQUE_LLM_GATEWAY_URL`.

**Session introductions, offline exercises, CPU training, and ComfyUI do not all require this endpoint configuration.** Follow the settings and prerequisites for the exercise you chose. The Claude apps use their documented deployment-specific setup.
