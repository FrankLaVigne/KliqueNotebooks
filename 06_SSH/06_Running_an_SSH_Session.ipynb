# 6 Running an SSH Session

**Purpose:** Work in a terminal on Klique-managed compute.

**What we're doing**

Launch an SSH Session, connect using its own dashboard instructions, inspect a repository, and end the session cleanly.

**Why we're doing it**

SSH is useful when your developer workflow is command-line driven. This lab makes the boundary between your local terminal and the remote shell explicit.

**What to look for**

* A connection command belonging to your instance
* A remote hostname and working directory
* A saved practice file and a stopped instance at the end

**Output:** A remote working copy and a recorded session ID.

## 6.1 Prerequisites and Mental Model

Have a Klique login, access to the SSH Session app and a suitable queue, an approved image with Git and Python 3 (some images provide `python3` but no `python` command, so this lab uses `python3`), and a local SSH client. Confirm required network access with your administrator. In your **local terminal**, check:

```text
ssh -V
```

If unavailable, arrange an approved SSH client before continuing.

```text
Local terminal → instance-provided SSH connection → remote shell and files
```

This is an SSH environment, not a browser editor or access to the control-plane server. The [Remote IDE guide](https://klique.ai/docs/latest/docs/getting_started/launching_remote_ide/) explains the managed-compute workflow.

## 6.2 Configure the Session

Open **Applications → AI Dev → SSH Session → Launch New**. The [SSH Session reference](https://klique.ai/docs/latest/docs/webapp/applications/apps_ssh_session) documents these defaults; your administrator may customize the form.

| Field | Lab choice |
|---|---|
| **Queue** | Approved queue with an assigned agent |
| **Container → Image** | Approved Python/Git image; check **Container requirements** |
| **Git → Repository / Branch** | Optional: this repository / `main` |
| **Store git repo** | Enable if using the Git form and preserving the checkout |
| **Persistent Workspace Path** | `~/workspace` |
| **Maximum idle time (in hours)** | Suggested: `1` |
| **Interactive Session Name** | `ssh-lab` |

For a non-root container, the reference requires **Use DropBear SSH server instead of OpenSSH server** under advanced options. Confirm your image's user with your administrator.

Submit with **Launch**, then inspect the instance's Console.

## 6.3 Connect from Your Local Terminal

When setup completes, the dashboard provides **Connect Command**, host/port, user, and password. Record the session ID.

Copy that instance's **Connect Command** into your local SSH-capable terminal. Follow its authentication instructions. Do not substitute the Klique website hostname, assume port 22, guess a username, or reuse another instance's details.

If SSH asks you to trust a new host key, verify the fingerprint through your administrator or approved process. Do not disable host-key checking to get past a warning.

After connecting, run these commands in the **remote shell**:

```bash
hostname
pwd
whoami
python3 --version
git --version
```

The session opens in the workspace the image already provides, so `pwd` prints `/root/workspace` and `whoami` prints `root`. The output looks like this, with your own hostname:

```
clearml-id-<session id>
/root/workspace
root
Python 3.10.12
git version 2.34.1
```

> **Checkpoint:** Record which hostname and directory belong to the remote shell. If authentication fails, return to the current dashboard rather than repeatedly trying old credentials.

## 6.4 Work with a Repository

If the launch form cloned the repo, enter that checkout. With snapshot storage selected, look under the configured workspace's `git_repo`; the unsynced alternative is a symlink, not a saved repository copy.

Otherwise clone it into the workspace. The image already has `~/workspace`, so there is nothing to create; `cd` there first in case you have moved. Run this in the **remote shell**, substituting your configured workspace if different:

```bash
cd ~/workspace
git clone https://github.com/FrankLaVigne/KliqueNotebooks.git
cd KliqueNotebooks
```

Skip cloning if a working copy already exists. For private repos, use approved Git authentication without embedding a token in the URL.

Inspect the project:

```bash
git status --short
git branch --show-current
git log -1 --oneline
ls
```

Create a small practice file outside the checkout but inside the persistent workspace:

```bash
python3 -c "from pathlib import Path; p = Path.home() / 'workspace' / 'ssh-lab-note.txt'; p.write_text('Created from the remote SSH session.\n'); print(p)"
cat ~/workspace/ssh-lab-note.txt
```

Adjust that path if your workspace differs. The command replaces that practice file if it already exists, so choose a new filename when keeping an earlier note.

> **Checkpoint:** The note prints successfully, and repository status remains unchanged.

## 6.5 Disconnect, Stop, and Restore

Save your work. Push intended source changes to a permitted Git remote if you need a Git backup; workspace persistence serves a different purpose.

Run `exit` to leave the SSH shell. Then return to Klique and use the instance action menu's **Stop** command. Disconnecting SSH alone is not proof that compute stopped. Verify the instance is stopped.

To revisit the files, clone the stopped instance's configuration or start another instance with **Restore Interactive Workspace ID** set to its session ID. Inspect the new session's connection details afresh, connect, and read the saved note. Stop this instance when finished too.

The [Remote IDE guide](https://klique.ai/docs/latest/docs/getting_started/launching_remote_ide/) describes file restoration into a fresh environment. Do not expect an old shell process to survive. See [instance actions](https://klique.ai/docs/latest/docs/webapp/applications/apps_overview#application-instance-actions) for Stop and Clone.

> **Checkpoint:** The saved note can be found after restoration, and no instance launched for this exercise remains running.

## Troubleshooting

| Symptom | Next check |
|---|---|
| Local `ssh` command missing | Arrange an approved client. |
| Setup stalls | Read Console; confirm queue capacity, image requirements, and Git access. |
| Connection times out | Check current instance state, exact host/port, and permitted network route. |
| Authentication rejected | Use current instance instructions; ask the administrator if those fail. |
| Host-key warning | Verify the host identity; do not bypass the warning. |
| Expected files absent | Check the configured workspace and chosen restore ID. |
| Session stops while you are away | Review idle settings and session logs. |
