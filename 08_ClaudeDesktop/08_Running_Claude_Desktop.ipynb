# 8 Running Claude Desktop

**Purpose:** Use Klique's Claude Desktop application through a remote desktop in your browser.

**What we're doing**

Launch the app, open the desktop, complete a small text exercise, review the result, and stop the instance.

**Why we're doing it**

Section 7 used a coding CLI inside VS Code. Here the interaction is the Claude Desktop application. Choose the interface that fits the work.

**What to look for**

* The correct application in the catalog
* A remote desktop containing Claude Desktop
* A useful result from a deliberately small task
* A stopped instance when finished

**Output:** A short reviewed checklist and notes identifying the session.

## 8.1 Prerequisites and Scope

Have a Klique WebApp login, access to Claude Desktop, an approved project and queue, and your organization's guidance on Claude account access. The app documentation does not specify the sign-in, subscription, API-key, or gateway configuration workflow; verify those locally rather than borrowing Claude Code's provider fields.

The [Claude Desktop reference](https://klique.ai/docs/latest/docs/webapp/applications/apps_claude_desktop) explicitly describes a **browser-accessible remote desktop** hosted on managed compute. This walkthrough does not install Claude Desktop on your laptop.

## 8.2 Configure and Launch

Select **Launch New** and review the default fields documented in the [app reference](https://klique.ai/docs/latest/docs/webapp/applications/apps_claude_desktop). Administrators may customize the form.

| Field | Suggested lab choice |
|---|---|
| **Compute Resource (Queue)** | Approved development queue |
| **Project** | Your permitted lab project |
| **Advanced Options → Session Name** | `claude-desktop-lab` |
| **Idle Shutdown (Hours)** | `1` |
| **Share Access** | Leave empty for this individual exercise |

Empty Share Access keeps visibility to you and tenant administrators. Leave environment variables and audio/microphone settings at approved defaults; this text exercise needs no audio.

Submit **Launch** and inspect **Console** during setup. Record the current session ID; once ready, use **Open Claude**.

## 8.3 Confirm the Interface and Access

Check that you see the desktop application in the remote desktop rather than a VS Code terminal. If an account or access prompt appears, follow your organization's approved process before continuing. Do not assume your local browser's Claude login carries into this environment.

If access is not configured, record the exact prompt for your administrator without sharing secrets. You can still inspect the Klique dashboard and stop the instance; do not leave it running while resolving account access.

> **Checkpoint:** The app is ready to accept a prompt, and you know which account/workspace you are using.

## 8.4 Try a Small Text Task

Paste this sample directly into a new conversation:

> Turn these fictional developer-session notes into a concise checklist: choose the approved compute queue; open the correct project; save changes before leaving; inspect Git status; record the session ID; stop the instance and verify it stopped. Use only these notes. Do not access files, connectors, or external services, and do not perform any actions.

Read the response. Does it preserve all six actions? Does it distinguish saving files from stopping compute? Ask it to fix any omission.

Copy the reviewed checklist to your own lab notes. No repository changes or upload are required.

The app also documents multi-file drag-and-drop uploads. If you explore that later, begin with a disposable text file and verify the destination and attachment in the interface before submitting; do not upload credentials or private repository contents as a first test.

> **Checkpoint:** You have a checklist you can compare directly with the six supplied notes.

## 8.6 Preserve Work and Stop

Copy any result you need into your normal durable storage before ending the session.

The app supports workspace persistence and **Continue from Previous Session**. Record the session ID for a future launch. The docs do not enumerate preserved directories or promise that authentication survives, so verify required files and access after a restore.

Use the instance action menu's **Stop** action and confirm stopped status. The **VNC Idle Disconnect Timeout** controls streaming after clients disconnect; do not confuse that with **Idle Shutdown** or use a closed browser tab as proof of cleanup.

For a later return, select the recorded session under **Continue from Previous Session**, review the launch settings, and launch. Stop that instance when finished too. General actions are documented in [Klique Applications](https://klique.ai/docs/latest/docs/webapp/applications/apps_overview#application-instance-actions).

> **Checkpoint:** Your checklist is saved outside the conversation as needed, the session ID is recorded, and the instance is stopped.

## Troubleshooting

| Symptom | Next step |
|---|---|
| App missing | Ask your administrator about catalog access. |
| Setup remains incomplete | Inspect Console and confirm queue availability. |
| Remote desktop does not open | Check instance status, link, and approved network access. |
| Claude requires sign-in or entitlement | Follow your organization's account process; this guide does not prescribe credentials. |
| Streaming stops | Check the instance state before reconnecting; a disconnected viewer and a stopped instance differ. |
| Restore does not bring back expected work | Check the previous-session selection and consult your deployment's persistence policy. |
| Governance behavior is unclear | Verify it with the administrator rather than applying Claude Code settings to this app. |
