# 5 Running VS Code

**Purpose:**
Open a browser-based development environment on Klique-managed compute and use it to work with this repository.

**What we're doing**

We launch the VS Code app, open the remote IDE, make a small local change, and finish by saving our work and stopping the session.

**Why we're doing it**

* Move the developer workflow alongside the compute used by the lab
* Practice the full lifecycle: start, work, save, stop, and return
* Keep source control and workspace persistence as separate habits

**What to look for**

* A working IDE link and a terminal running on the remote machine
* The repository's files in the editor
* A saved change that you can find after restoring a workspace

**Output**

A remote working copy of KliqueNotebooks, a small practice script, and a recorded session ID for a later restore.

> **Before you start:** You need a Klique WebApp login with access to the VS Code app, an available queue serviced by an orchestrator, and an approved container environment with Python and Git. Ask your administrator which queue and image to use. Editing this lab does not require a model endpoint or a GPU workload.

This is a guided UI lab. Run the fenced shell commands in the **remote IDE terminal**, not in your laptop's terminal or as Python notebook cells.

## 5.1 Where the IDE Runs

The browser displays the editor; the development environment runs on Klique-managed compute. The orchestrator serving your selected queue starts the IDE on its machine. This does not mean installing VS Code on the Klique control-plane server.

```text
Your browser → remote VS Code session → repository, terminal, Python
                        ↑
              managed compute selected by queue
```

The endpoint work in Sections 1–4 and this developer workspace have separate lifecycles. Stopping this IDE is not the same as stopping a model deployment.

See [Remote Development Environment](https://klique.ai/docs/latest/docs/orchestration_engine/gpuaas_remote_ides) and [Launching a Remote IDE](https://klique.ai/docs/latest/docs/getting_started/launching_remote_ide/).

## 5.2 Configure and Launch

Open **Applications → AI Dev → VS Code**, select **Launch New**, and configure the instance. Submit with **Launch**. These actions are documented in [Klique Applications](https://klique.ai/docs/latest/docs/webapp/applications/apps_overview).

The [VS Code app reference](https://klique.ai/docs/latest/docs/webapp/applications/apps_vscode) describes the default fields below; administrators can customize them.

| Field | Suggested lab choice |
|---|---|
| **Queue** | Your administrator's development queue |
| **Container → Image** | Approved Python/Git image; check the app's **Container requirements** |
| **Git → Repository** | `https://github.com/FrankLaVigne/KliqueNotebooks.git` |
| **Git → Branch** | `main` |
| **Store git repo** | Enable to include the repository in the snapshot under `./git_repo` |
| **Persistent Workspace Path** | `~/workspace` |
| **Maximum idle time (hours)** | `1` for this lab |
| **Interactive Session Name** | `vscode-lab` |
| **Advanced Options → VSCode additional extensions** | Optional: `ms-toolsai.jupyter,ms-python.python` |

The one-hour timeout is a lab suggestion. Advanced options define idleness using network, CPU, and GPU thresholds.

> **Checkpoint:** Record your queue, image, and workspace path before launching. Do not substitute the vLLM image from Section 1 without checking IDE requirements.

## 5.3 Open the Session

Select the new instance. Follow setup in **Console**; when ready, use **Open IDE**. The dashboard also shows status, resource usage, and the current session ID. Record that ID.

In the IDE, open a terminal and run:

```bash
pwd
hostname
python --version
git --version
```

These are inspection commands. Compare the hostname and working directory with your local machine to establish where your commands are executing.

> **Checkpoint:** The editor opens and the terminal reports Python and Git versions. If either command is missing, fix the approved environment before continuing.

## 5.4 Find or Clone the Repository

If the launch form cloned the repository, open that folder in the IDE and use its terminal. With **Store git repo** enabled, look for `git_repo` inside your configured workspace; without it, the documented snapshot entry is a `git_repo_not_synced` symlink. Do not assume that symlink preserves repository contents.

If you left the launch form's Git settings empty, clone manually into your persistent workspace. The example below assumes you chose `~/workspace`; change it if you configured another location. Skip cloning if you already have a working copy.

```bash
mkdir -p ~/workspace
cd ~/workspace
git clone https://github.com/FrankLaVigne/KliqueNotebooks.git
cd KliqueNotebooks
```

For a private repository, use your organization's approved Git authentication. Never put a credential into the clone URL or notebook.

From the repository root, inspect the checkout:

```bash
git remote -v
git branch --show-current
git status --short
```

Open `README.md` and this notebook, `05_VSCode/05_Running_VS_Code.ipynb`, in the editor.

> **Checkpoint:** You can see `config.py` and the numbered section directories. Use the existing checkout rather than creating a second copy.

## 5.5 Make a Small Change

For this exercise, create a local branch from the repository terminal:

```bash
git switch -c lab/vscode-practice
```

If that branch already exists, inspect it and switch to it instead. In the editor, create `hello_remote.py` at the repository root:

```python
import platform
from pathlib import Path

print(f"Hello from {platform.node()}")
print(f"Working directory: {Path.cwd()}")
```

Save the file, then run:

```bash
python hello_remote.py
git status --short
```

Expected: a greeting, a working directory, and an untracked `hello_remote.py` entry. Edit the greeting, save again, and rerun it.

You can now use this workspace for earlier lab sections. Install their requirements in the remote environment, select its Python kernel when opening notebooks, and recreate the ignored `.env` there using Section 1.6. A fresh clone does not include your laptop's packages or secrets.

> **Checkpoint:** Your changed greeting appears. No model call is needed for this exercise.

## 5.6 Save, Stop, and Return

Before stopping, save editor buffers and inspect `git status`. Commit only the files you intend to keep; push to your own permitted repository or branch if you need a durable Git copy. A local commit alone is not a remote backup. This exercise does not require pushing the practice script to the shared lab repository.

Klique's persistent workspace captures files for restoration when a session closes. Keep the practice file under the configured workspace root. A later session restores those files into a **fresh environment**, rather than resuming the old process. See [Launching a Remote IDE](https://klique.ai/docs/latest/docs/getting_started/launching_remote_ide/).

1. Return to the VS Code instance list.
2. Open your instance's action menu, or right-click it, and select **Stop**.
3. Verify the instance is stopped. Closing the browser tab is not a cleanup check.
4. To practice returning, use **Clone** on that instance, review the prefilled configuration, and launch a new session. Alternatively, set **Restore Interactive Workspace ID** to the recorded ID in a new launch form.
5. Open the new IDE, find `hello_remote.py`, and run it again. Then stop the new instance too.

The configured idle timeout provides automatic shutdown if the session becomes idle. Do not use **Delete** as a substitute for this stop-and-restore exercise. For action semantics, see [Application Instance Actions](https://klique.ai/docs/latest/docs/webapp/applications/apps_overview#application-instance-actions).

> **Checkpoint:** Your saved greeting is available in the restored workspace, and your final instance is stopped.

## Troubleshooting

| Symptom | What to check |
|---|---|
| VS Code app or queue unavailable | Ask your administrator about access and the appropriate development queue. |
| Setup does not finish | Read Console for image, package, Git authentication, or queue problems. |
| IDE link does not open | Check instance status and your organization's network/access requirements; use the dashboard's link. |
| Notebook has no usable Python kernel | Check the remote Python environment and notebook extensions. |
| Changes missing after restore | Confirm the session ID, workspace root, and whether repository contents were included in the snapshot. |
| Session ends unexpectedly | Review the configured idle timeout and idleness thresholds. |
