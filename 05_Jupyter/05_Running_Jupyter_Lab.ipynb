## 5.1 Launch the Hosted Workspace

1. In the Klique WebApp, open **Applications → AI Dev → Jupyter Lab** and choose **Launch New**.
2. Select your approved **Queue**, serviced by a Klique Orchestrator, and the compatible **Container → Image**. Do not assume the image used for vLLM is also a notebook image.
3. Set **Interactive Session Name**, for example `jupyter-first-lesson`. Check **Advanced Options → Interactive Session Project** and select the project your administrator approved.
4. Set **Persistent Workspace Path** to the approved location, for example `~/workspace`. Confirm an appropriate **Maximum idle time (hours)**. These are deployment choices, not universal defaults.
5. If your deployment offers Git settings, use this repository and `main`. Confirm whether **Store git repo** is enabled before relying on repository snapshots. Otherwise, use the JupyterLab terminal to clone into the persistent workspace after launch. Reuse an existing checkout rather than overwriting work.
6. Choose **Launch**, wait for setup, and inspect **Console** if it stalls. When ready, use the instance dashboard's **Open IDE** link. Record the instance/session ID for a future restore.

The browser displays JupyterLab; the kernel and terminal run on the selected managed compute, not your laptop. Use the dashboard's authenticated link rather than publishing a new public server.

## 5.2 Open a Notebook and Choose a Kernel

Open `05_Jupyter/05_Running_Jupyter_Lab.ipynb` from your checkout. Select the approved **Python 3** kernel when prompted. For a separate practice notebook, use the **+** Launcher, choose the Python kernel, and rename the new notebook `first_notebook.ipynb` inside the persistent workspace.

A **Markdown cell** contains explanatory text, headings, and lists. A **code cell** sends Python to the selected kernel. Run a selected cell with **Shift+Enter**. The kernel holds variables in memory until it restarts or shuts down; the notebook file stores source and any saved outputs.

Run the next three cells from top to bottom. No dependency installation is needed. Do not run these as shell commands.


```python
message = "Hello from Klique"
print(message)
```

```python
numbers = [2, 4, 6]
total = sum(numbers)
print(f"Total: {total}")
```

```python
print(f"{message}. The total is {total}.")
```

**Expected:** the first cell prints a greeting, the second prints `Total: 12`, and the third combines them. Change a number, rerun the second and third cells, and observe the new total.

Save the notebook. Then restart the kernel using JupyterLab's **Kernel** menu. Run only the third cell first: it should raise `NameError` because the variables are no longer in memory. Run all three cells from the top to restore the result, then save again. A saved output is not proof that the current kernel has run its source.

## 5.3 Dependencies and Persistence

This first lesson uses Python's standard library only. Later endpoint notebooks need `requests`; other labs have their own requirements. Use the approved image/package source for that lab instead of installing every dependency here. With an IPython Python kernel, `%pip install requests` targets that kernel's environment, but only run it when package installation is permitted and needed. Restart the kernel if the package instructions require it.

In offline or air-gapped deployments, packages must be preinstalled or available from an approved reachable package repository. A package installed in a running session may need to be installed again in a fresh environment. Record the image and dependency versions needed to reproduce your work.

Save `.ipynb` files under the configured persistent workspace, and confirm whether the checkout itself is included in snapshots. Snapshots restore files, not running kernels, variables, terminals, or training processes. Do not assume files elsewhere are preserved. A Git commit on the same remote machine is not an independent backup.

Keep credentials out of cells, outputs, screenshots, and Git. Ignored files can still be included in workspace snapshots.

## 5.4 Save, Stop, and Return

Save the notebook before leaving. Closing its browser tab does **not** shut down its kernel or the Klique session. JupyterLab's **Running** panel can shut down kernels and terminals, but that alone is not a check that managed compute has stopped.

Return to the Klique instance controls, choose **Stop**, and verify the stopped status. A separately launched model deployment has its own lifecycle. Do not use **Delete** as a substitute for the restore exercise.
