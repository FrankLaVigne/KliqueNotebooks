# 7 Running Claude Code

**Purpose:** Use Klique's managed coding-agent environment for a small, reviewable developer task.

**What we're doing**

Launch Claude Code, confirm the model route, open the remote editor, ask for a small local change, inspect it, and stop the session.

**Why we're doing it**

The development loop now includes an assistant. You still define scope, review changes, and decide what is worth keeping.

**What to look for**

* A configured gateway profile and its intended model
* A small change you can inspect with Git
* Usage recorded during the exercise

**Output:** A reviewed practice script and session notes recording the model route and workspace.

## 7.1 Prerequisites

You need a Klique login, access to the Claude Code app, an approved project and compute queue, and administrator guidance on the gateway profile/model and provider setup. Confirm those before launching. This lab does not require supplying a personal API key.

Klique's Claude Code app supplies a remote VS Code environment with the CLI already configured. It is distinct from the Claude Desktop application in Section 8. See the [Claude Code app](https://klique.ai/docs/latest/docs/webapp/applications/apps_claude_code/).

Keep the exercise inside a disposable project folder. The session may have tools that can affect platform resources, so the first prompt below explicitly limits its scope.

## 7.2 Choose the Route and Launch

Open **Claude Code → Launch New**. Choose your **Compute Resource (Queue)** and **Claude Code Session Project**. Under advanced options, name the session `claude-code-lab`, choose an approved **Container Language**, keep **Workspace Autosave** enabled, and set **Idle Shutdown (Hours)** to a suggested `1`. Review **Persistent Workspace Path**.

The documentation describes the provider route in two ways:

* The [agentic session guide](https://klique.ai/docs/latest/docs/getting_started/launching_agentic_session) describes selecting a gateway **Profile** and using that profile's default model.
* The [app reference](https://klique.ai/docs/latest/docs/webapp/applications/apps_claude_code/) describes **AI Provider → Vendor Endpoint** for gateway routing; blank means direct Anthropic access. A custom key is optional, with a User Vault default.

Use the controls available in your deployment and confirm the intended gateway profile/default model with your administrator. Do not infer a profile selector or endpoint value that your form does not show. Record the route and model before submitting **Launch**.

> **Checkpoint:** You know whether calls will use the gateway or go directly to a provider. This lab's governance comparison requires the gateway route.

## 7.3 Open the Coding Session

Watch **Console** during setup. Record the session ID and use **Open Claude** to enter the browser VS Code session.

In its terminal, inspect the environment:

```bash
pwd
python --version
git --version
claude --version
```

Use a checkout already present, or clone this repository under your configured persistent path. The following assumes the workspace includes your home directory:

```bash
mkdir -p ~/workspace
cd ~/workspace
git clone https://github.com/FrankLaVigne/KliqueNotebooks.git
cd KliqueNotebooks
git status --short
git switch -c lab/claude-code-practice
```

Skip cloning if the repo exists; inspect and reuse the practice branch if it already exists. Do not overwrite someone else's pending changes.

> **Checkpoint:** You are in the repository root, you understand the current Git status, and the CLI reports a version.

## 7.4 Give the Agent a Bounded First Task

Start the configured CLI from the repository terminal with `claude` if it is not already open. Use this prompt:

> Work only in this local repository. Read README.md to understand the lab. Create one new file, hello_klique.py, using only the Python standard library. It should print a short greeting and the current working directory. Do not read .env or credentials, edit any existing file, use Klique MCP tools, launch resources, install packages, commit, or push. Explain the proposed file before editing and ask before executing commands.

Read its proposal and any permission prompts. Approve only the scoped practice change. If it proposes platform actions or broader edits, stop and narrow the task.

In a separate terminal, inspect the result:

```bash
git status --short
git diff --stat
git diff
```

A new untracked file is not shown by plain `git diff`; open `hello_klique.py` in the editor and review it directly. After review, run:

```bash
python hello_klique.py
```

Expected: a greeting and the repository directory. No model deployment or external service is needed by the generated script.

> **Checkpoint:** Only the intended new file appears, its contents are understandable, and its output matches the request.

## 7.5 Inspect Usage

Return to the app dashboard and inspect the token charts. Klique's outbound proxy meters session usage; gateway routing additionally exposes it in **AI Usage** and supports quota enforcement. See the [agentic session guide](https://klique.ai/docs/latest/docs/getting_started/launching_agentic_session).

Record the time window and visible input/output totals before and after another small explanation request. Do not assume one prompt equals one model call.

If you have access to AI Usage, compare the corresponding window there. If usage appears only in the app dashboard, verify the provider route before concluding that gateway governance applied.

> **Checkpoint:** You can identify where your session's usage is visible and explain which route produced it.

## 7.6 Save and Stop

Save buffers, inspect Git status, and keep or discard the practice file intentionally. Push only intended changes to your permitted remote if you need a Git backup.

The persistent path defaults to `~/`, including dotfiles, cloned repositories and `~/.claude`; a blank path makes the session ephemeral. Restoration creates a fresh container, not a resumed process. Use **Continue from Previous Session** when launching a later session. See the [agentic session guide](https://klique.ai/docs/latest/docs/getting_started/launching_agentic_session).

Treat workspace snapshots as potentially containing agent history and credentials. Do not share them as if they were source-only exports.

Exit the CLI, return to Klique, and choose **Stop** from your instance's action menu. Verify stopped status; closing the browser is not sufficient. Stop any restored practice session too.

> **Checkpoint:** You recorded the session ID and workspace path, and the instance is stopped.

## Troubleshooting

| Symptom | What to check |
|---|---|
| App or project unavailable | Verify access with your administrator. |
| Setup fails | Read Console and check queue/image availability. |
| CLI asks for unexpected authentication | Recheck the configured provider route and approved vault setup; do not paste a key into chat. |
| Wrong model or missing AI Usage entry | Verify profile default model or Vendor Endpoint against this deployment's configuration. |
| Quota/rate rejection | Inspect the applicable gateway policy; do not bypass it with a direct provider key. |
| Practice change is too broad | Stop the agent, inspect Git status, and narrow the prompt. |
| Files missing later | Check persistent path and selected previous session. |
